### Dictionary Assignment

```python
# Question 1

d = {1: 1, 2: 2, 3: 3, 4: 4}

sum_keys = sum(d.keys())

print("Dictionary:", d)
print("Sum of Keys:", sum_keys)
```

```python
# Question 2

d = {1: 1, 2: 2, 3: 3, 4: 4}

sum_values = sum(d.values())

print("Dictionary:", d)
print("Sum of Values:", sum_values)
```

```python
# Question 3

d = {1: 1, 2: 2, 3: 3, 4: 4}

total = sum(d.keys()) + sum(d.values())

print("Dictionary:", d)
print("Total of Keys and Values:", total)
```

```python
# Question 4

user_data = {}

while True:
    key = input("Enter Key: ")
    value = input("Enter Value: ")

    user_data[key] = value

    choice = input("Do you want to add more? (yes/no): ")

    if choice.lower() != "yes":
        break

print("\nFinal Dictionary:")
print(user_data)
```

```python
# Question 5

student_score = {1: 44, 2: 45, 3: 55}

total_score = sum(student_score.values())

print("Student Scores:", student_score)
print("Total Score:", total_score)
```

```python
# Question 6

d = {1: "A", 2: "B", 3: "C", 4: "D", 5: "E", 6: "F"}

odd_keys = {}
even_keys = {}

for key, value in d.items():
    if key % 2 == 0:
        even_keys[key] = value
    else:
        odd_keys[key] = value

print("Odd Keys Dictionary:", odd_keys)
print("Even Keys Dictionary:", even_keys)

print("Total Odd Keys:", len(odd_keys))
print("Total Even Keys:", len(even_keys))
```

```python
# Question 7

player = {
    7: "Dhoni",
    12: "Kohli",
    9: "Rohit",
    89: "Bumrah"
}

greatest_key = max(player.keys())

print("Dictionary:", player)
print("Greatest Key:", greatest_key)
print("Player:", player[greatest_key])
```

```python
# Question 8

player = {
    7: "Dhoni",
    12: "Kohli",
    9: "Rohit",
    89: "Bumrah"
}

items = list(player.items())

alternate = dict(items[::2])

print("Original Dictionary:", player)
print("Alternate Key-Value Pairs:", alternate)
```

```python
# Question 9

player = {
    7: "Dhoni",
    12: "Kohli",
    9: "Rohit",
    89: "Bumrah"
}

result = []

for value in player.values():
    if value.startswith("K"):
        result.append(value)

print("Values Starting with K:", result)
```

```python
# Question 16

text = "banana"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("String:", text)
print("Character Frequency:", frequency)
```

```python
# Question 20

student_marks = {
    "A": 45,
    "B": 70,
    "C": 90,
    "D": 35,
    "E": 65
}

count = 0

for value in student_marks.values():
    if value > 50:
        count += 1

print("Dictionary:", student_marks)
print("Values greater than 50:", count)
```

```python
# Question 21

marks = {
    "A": 85,
    "B": 90,
    "C": 75,
    "D": 95
}

highest_key = max(marks, key=marks.get)

print("Dictionary:", marks)
print("Key with Highest Value:", highest_key)
print("Highest Value:", marks[highest_key])
```

```python
# Question 22

student = {
    1: "Rahul",
    2: "Aman",
    3: "Priya"
}

key = int(input("Enter key: "))
value = input("Enter value: ")

if key in student:
    student[key] = value
    print("Key found. Value updated.")
else:
    student[key] = value
    print("Key not found. New key added.")

print("Updated Dictionary:", student)
```

```python
# Question 23

keys = [1, 2, 3]
values = ["a", "b", "c"]

result = dict(zip(keys, values))

print("Keys:", keys)
print("Values:", values)
print("Dictionary:", result)
```

```python
# Question 24

d = {
    1: "apple",
    2: "banana",
    3: "apple",
    4: "orange",
    5: "banana"
}

unique_dict = {}

for key, value in d.items():
    if value not in unique_dict.values():
        unique_dict[key] = value

print("Original Dictionary:")
print(d)

print("\nDictionary after removing duplicate values:")
print(unique_dict)
```

```python

# Question 25

d = {
    1: "apple",
    2: "banana",
    3: "orange",
    4: "grapes"
}

if len(d.values()) == len(set(d.values())):
    print("All values are unique.")
else:
    print("Duplicate values found.")
```
