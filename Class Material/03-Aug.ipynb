```python
#### Asynchronous Programming
import asyncio
async def my_task(name,delay):
    print(f"{name} started")
    await asyncio.sleep(delay)
    print(f"{name} done after {delay}")
    
async def main():
    await asyncio.gather(
        my_task("Registration - A",10),
        my_task("Login - B",1),
        my_task("Add Data - C",3),
        my_task("Logout - D", 0.5)
        
    )
    
await main()
```

```python
# with sync style code
import requests 
def task(name,endpoint):
    print(f"{name} started")
    resp = requests.get(f"https://dummyjson.com/{endpoint}")
    data = resp.json()
    print(f"{name} done with -> status {resp.status_code}")
    return data
 
def main():
    results  = [     
    task("User Tasks","users/1"),
    task("Product Tasks","products/1"),
    task("Recipe Task","recipes/1")
    ]
    print("\n All result collected")
    for r in results:
        print(r)
main()
```

```python
## with async style
import asyncio
import aiohttp

async def task(session,name, endpoint):
    print(f"{name} started")
    async with session.get(f"https://dummyjson.com/{endpoint}") as resp:
        data = await resp.json()
        print(f"{name} done with -> status {resp.status}")
        return data
    
async def main():
    async with aiohttp.ClientSession() as session:
        results = await asyncio.gather(
            task(session,"User Tasks","users/1"),
            task(session,"Product Tasks","products/1"),
            task(session,"Recipe Task","recipes/1")
        )
    print("\n All result collected")
    for r in results:
        print(r)
await main()
```
