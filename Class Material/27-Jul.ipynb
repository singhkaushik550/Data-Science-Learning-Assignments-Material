```python
class Student:
    def __init__(self):
        self.students = {}

    def add_students(self, roll_no, name, age, city):
        if roll_no in self.students:
            print("Student already exists")
            return

        self.students[roll_no] = {
            "name": name,
            "age": age,
            "city": city
        }

        print(f"Student {name} added successfully.")

    def get_all_students(self):
        if not self.students:
            print("No Students Found")
            return

        for roll_no, details in self.students.items():
            print(f"""
Roll No : {roll_no}
Name    : {details['name']}
Age     : {details['age']}
City    : {details['city']}
----------------------------
""")

    # Get Student by Roll Number
    def get_student_by_roll(self, roll_no):
        if roll_no in self.students:
            details = self.students[roll_no]

            print(f"""
Roll No : {roll_no}
Name    : {details['name']}
Age     : {details['age']}
City    : {details['city']}
""")
        else:
            print("Student not found.")

    # Update Student
    def update_student(self, roll_no, name, age, city):
        if roll_no not in self.students:
            print("Student not found.")
            return

        self.students[roll_no] = {
            "name": name,
            "age": age,
            "city": city
        }

        print("Student updated successfully.")

    # Delete Student
    def delete_student(self, roll_no):
        if roll_no in self.students:
            del self.students[roll_no]
            print("Student deleted successfully.")
        else:
            print("Student not found.")


# Object
s = Student()

s.add_students(100, "Mike", 44, "Noida")
s.add_students(101, "Sachin", 22, "Delhi")

print("All Students")
s.get_all_students()

print("Get Student")
s.get_student_by_roll(100)

print("Update Student")
s.update_student(100, "Mike", 45, "Gurgaon")
s.get_student_by_roll(100)

print("Delete Student")
s.delete_student(101)

print("After Delete")
s.get_all_students()

```

```python
 # Inheritance 
 # When one class (child/serived) reuses code and functionality from
 # another class(parent/base) known as inheritance.
 #1. Single Inheritance
 #2. Multilevel Inheritance
 #3. Multiple Inheritance
 #4. Herirchal Inheritance
 #5. Hybrid
```

```python
#1. Single Inheritance
class person:
    def show(self):
        print("I am a person")

class student(person):
    pass
stu = student()
stu.show()
```

```python
#2. Multilevel Inheritance
class person:
    def show(self):
        print("I am a person")

class student(person):
    def study(self):
        print("I got rank 1")
        
class TopperStudent(student):
    def rank(self):
        print("I got rank 1")
        
        
t= TopperStudent()
t.show()
t.study()
t.rank()
```

```python
#3. Multiple Inheritance
class father:
    def skills(self):
        print("Coding")
class mother:
    def talent(self):
        print("Drawing")
        
class child(father,mother):
    pass

c= child()

c.skills()
c.talent()
```

```python
 #4. Herirchal Inheritance
class person:
    def show(self):
        print("I am a person.")
        
class student(person):
    pass

class teacher(person):
    pass

class driver(person):
    pass

class painter(person):
    pass

student().show()
teacher().show()
driver().show()
painter().show()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name = name # public member
        self.__marks = marks ## private member( if we will do it in private it will not changed)
        
    def student_result(self):
        print(self.name)
        print(self.__marks)
        
    def set_marks(self,marks):
        if 0<= marks<=100:
            self.__marks = marks
        else:
            print("invalid marks")


s = Student("Dev",66)
s.student_result()    
```

```python
## Abstraction
from abc import ABC, abstractmethod
class payment(ABC):
    @abstractmethod
    def pay(self,amount):
        print(f"Paid {amount} using credit card")
        
class credit_card_payment(payment):
    def pay(self,amount):
        print(f"Paid {amount} using credit card")

class upi_payment(payment):
    def pay(self,amount):
        print(f"Paid {amount} using UPI")
p = upi_payment()
p.pay(300)
```

```python
p = credit_card_payment()
p.pay(1202002)
```

```python
from abc import ABC, abstractmethod

class rbi_bank(ABC):
    def details(self):
        print("Your bank connevcted to RBI bank")
        
    def database(self):
        print("Databse is activated for RBI bank")
    
    @abstractmethod    
    def security(self):
        pass
    
class sbi(rbi_bank):
    def details(self):
        super().details()
        print("Welcome to sbi")
        
    def security(self):
        print("SBI security OTP + PIN")
```

```python
# Abstract class = A class you can not crfeate objects from directly 
# it is a blue print /templet for other class.
# # Abstract method = A method 
# abstractmethod
## Abstract 
sbi_user = sbi()
sbi_user.details()
sbi_user.database()
```
