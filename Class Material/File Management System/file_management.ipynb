```python
############### File Managemet System ####################

import os
class FileManagementSystem():
    def __init__(self):
        self.folder = "FMS"
        
        if not os.path.exists(self.folder):
            os.mkdir(self.folder)
    
    def create_file(self,file_name,content):
        path = os.path.join(self.folder,file_name)
        
        if os.path.exists(path):
            return "File already exists."

        with open(path, "w") as file:
            file.write(content)

        return "File created successfully."
    
    def update_file(self,file_name,new_content):
        path = os.path.join(self.folder,file_name)
        
        if not os.path.exists(path):
            return "Files not found"
        
        with open(path,"a") as file:
            file.write("\n"+new_content)
            
        return "File Updtaed successfully."
        
    def read_file(self):
        files = os.listdir(self.folder) 
        
        if not files:
            print("File not found")
            
        for file_name in files:
            print(file_name)
            path = os.path.join(self.folder,file_name)
            
            with open(path,"r") as file:
                print(file.read())
                
    def delete_file(self,file_name):
        path = os.path.join(self.folder,file_name)
        
        if not os.path.exists(path):
            return "File not found"
        
        os.remove(path)
        return "Files removed successfully"            
        
fms = FileManagementSystem()
file_name = "trial.txt"
content = "This is the trial system"
fms.create_file(file_name,content)

new_content = "This is the second line"
fms.update_file(file_name,new_content)

fms.read_file()

fms.delete_file(file_name)
```
