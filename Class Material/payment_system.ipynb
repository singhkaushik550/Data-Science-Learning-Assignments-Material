```python
############## Payment System ################
from abc import ABC, abstractmethod
```

```python
class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass
```

```python
class credit_card(Payment):

    def __init__(self, card_number):
        self.card_number = card_number

    def process_payment(self, amount):
        print("\n Credit Card Payment")
        print(f"Card Number : {self.card_number}")
        print(f"Amount Paid : ₹{amount}")
        print("Payment Successful")
```

```python
class upi_payment(Payment):
    def __init__(self,upi_id):
        self.upi_id = upi_id
        
    def process_payment(self, amount):
        print("\n UPI Payment")
        print(f"UPI ID: {self.upi_id}")
        print(f"Amount Paid: ₹{amount}")
        print("Payment Successful")
```

```python
class wallet_payment(Payment):
    def __init__(self,balance):
        self.balance = balance
        
    def process_payment(self, amount):
        if amount>self.balance:
            raise ValueError("Insufficient Balance")
        
        self.balance -= amount
        print("\n Wallet Payment")
        print(f"Amount Paid: ₹{amount}")
        print(f"Remainning Balance: ₹{self.balance}")
        print("Payment Successful")
```

```python
class checkout:

    def __init__(self, payment_method):
        self.payment_method = payment_method

    def pay(self, amount):
        self.payment_method.process_payment(amount)
```

```python
credit = credit_card("1234-5678-9876-1111")
checkout = checkout(credit)
checkout.pay(2500)
```

```python
upi = upi_payment("7982979700@axl")
checkout = checkout(upi)
checkout.pay(10000)
```

```python
class netbanking(Payment):
    def __init__(self,bank_name):
        self.bank_name = bank_name
        
    def process_payment(self, amount):
        print("\n Net Banking Payment")
        print(f"Bank Name: {self.bank_name}")
        print(f"Amount Paid: ₹{amount}")
        print("Payment Successful")
```

```python
net = netbanking("State Bank of India")
checkout = checkout(net)
checkout.pay(12000)
```
