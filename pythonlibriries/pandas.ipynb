```python
# valid email generator
import re

email = "adam@gmail.com"
pattern = r"^[\w.-]+@[\w.]+\.\w"

if re.match(pattern,email):
    print("Valid Email")
else:
    print("Invalid Email")
```

```python
# password generator
import re
password = "Adam@123"
pattern = r"^(.[A-Z])(.[a-z])(.[\d])(.[@#!%$&]).{8}"

if re.match(pattern,password):
    print("valid Password")
else:
    print("Invalid Password")
```
